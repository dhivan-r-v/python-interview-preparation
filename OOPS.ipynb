# Note:
* Self: means creating or using the current object
* __init__ : means constructor,it runs automatically when the object is created
* Why is self everywhere? : If we don't use self python does not know whether to update Dhivan's or Vignesh's data, that's why self is important if we self it searches for current object and update the input we have given

```python
class Secure:                                
    def __init__(self,name,atm_pin,balance):
        self.name=name
        self.__atm_pin=atm_pin  #Encapsulation
        self.balance=balance

    def check_pin(self,attempt):
        return attempt==self.__atm_pin

    def dep_cash(self,amount):
        self.balance+=amount
        print(f"{amount} deposited")

    def cashwithdrawl(self,amount):
        if amount<=self.balance:
            self.balance-=amount
            print(f"{amount} withdrwan")
        else:
            print("Insufficient Balance")

    def transaction(self):
        print("Per day Transaction limit is 10000")
        
class PremiumUser(Secure):   #Inheritence
    def cashwithdrawl(self,amount):
        if amount<=self.balance+10000:
            self.balance-=amount
            print("You got special withdrawl")
        else:
            print("Limit Exceeded")

    def transaction(self):         #Here the transaction acts as different under two classes(Polymorphism)
        print(" Per day Transaction limit is 1,00,000")

    

        
```

```python
Dhivan=Secure("Dhivan",2000,50000)
Dhivan.dep_cash(75000)
Dhivan.cashwithdrawl(25000)
Dhivan.transaction()

Vignesh=PremiumUser("Vignesh",1998,250000)
Vignesh.dep_cash(125000)
Vignesh.cashwithdrawl(80000)
Vignesh.transaction()
```
