```python
Even_list=[]
for i in range(1,50):
    if i%2==0:
        Even_list.append(i)
print("The Even Numbers are",Even_list)
```

### 3.	Write a program that handles division by zero using try-except.

```python
Numerator=int(input("Enter the numerator"))
Denominator=int(input("Enter the denominator"))
try:
    if Numerator%Denominator==0:
        print("No Remainder")
except ZeroDivisionError:
    print("Enter values other than Zero")
```

### 4.	Given a string, count how many vowels it has

```python
sentence=input("Enter word/sentence:").lower()
count=0
for character in sentence:
    if character in "aeiou":
        count+=1
print("The total no of vowels are",count)
```

### 5.	Reverse a list without using built-in functions.

```python
list=[1,2,3,4,5]
i=len(list)-1
reverse=[]
while i>=0:
    reverse.append(list[i])
    i=i-1
print(reverse)
    
```
