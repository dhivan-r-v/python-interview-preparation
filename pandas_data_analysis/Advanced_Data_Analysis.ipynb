### 1.2 Pull up the avg insurance based on region and sex

### Discrete and continuous data
* Discrete/categorical data - countable/can't be segmented
* Continuous - Measurable and have units/can also be segmented

```python
insurance_data.groupby(by=["region","sex"])["charges"].max().sort_values(ascending=False).round(2) #CHAIN METHODS
```

### 6.1 Create a new column in the same dataframe with the name "Eligibility Criteria" where if age is > 20,say Eligible else not

```python
def get_age(x):
    if x>20:
        return "Eligible"
    else:
        return "Not Eligible"
```

```python
insurance_data["Eligibility Criteria"]=insurance_data["age"].apply(get_age)
insurance_data
```

## 7.lambda function

* Anonymous function/nameless function

```python
insurance_data["EC"]=insurance_data["age"].apply(lambda x:"ELigible" if x>20 else "NE")
insurance_data
```
