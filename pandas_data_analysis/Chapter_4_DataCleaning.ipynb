# Task1 Data Cleaning

```python
import pandas as pd
sales_data_2017=pd.read_csv("Sales Transaction 2017.csv")
sales_data_2018=pd.read_csv("Sales Transaction 2018.csv")
sales_data_2019=pd.read_csv("Sales Transaction 2019.csv")

sales_full_list=pd.concat(objs=[sales_data_2017,sales_data_2018,sales_data_2019])

sales_full_list["Date"]=sales_full_list["Date"].str.strip() 
sales_full_list=sales_full_list.dropna(subset=["Date"]) 
sales_full_list["Date"]=pd.to_datetime(sales_full_list["Date"],dayfirst=True)

sales_full_list=sales_full_list.dropna(subset=["Voucher"])
sales_full_list=sales_full_list[sales_full_list["Voucher"].str.contains("Sal:",na=False)]
sales_full_list["Voucher"]=sales_full_list["Voucher"].str.strip().str.replace("Sal:","",regex=False).astype(int)
sales_full_list["Qty"]=sales_full_list["Qty"].str.strip().str.replace(",","",regex=False).astype(float).astype(int)
sales_full_list=sales_full_list.dropna(subset=["Rate"])
sales_full_list["Rate"]=sales_full_list["Rate"].str.strip().str.replace(",","",regex=False).astype(float)
sales_full_list=sales_full_list.dropna(subset=["Gross"])
sales_full_list["Gross"]=sales_full_list["Gross"].str.strip().str.replace(",","",regex=False).astype(float)
sales_full_list["Disc"]=sales_full_list["Disc"].fillna("0")
sales_full_list["Disc"]=sales_full_list["Disc"].str.strip().str.replace(",","",regex=False).astype(float)
sales_full_list["Voucher Amount"]=sales_full_list["Voucher Amount"].fillna("0")
sales_full_list["Voucher Amount"]=sales_full_list["Voucher Amount"].str.strip().str.replace(",","",regex=False).astype(float)
sales_full_list



```

# Task 2 - Merge Function 1 Example

## Inner Join -- (if how is not used it automatically consider it as inner join)

```python
import pandas as pd
employee_data=pd.read_csv("employee_data.csv")
empdata=pd.read_csv("empdata.csv")
pd.merge(employee_data,empdata,on="Name")
```

## Right Join

```python
pd.merge(employee_data,empdata,on="Age",how="right")
```

## Left Join

```python
pd.merge(employee_data,empdata,on="Age",how="left")
```

# Task 4 -- Lambda Function 2 examples with Map, Filter, Reduce


```python
import pandas as pd
insurance_details=pd.read_csv("insurance.csv")
insurance_details
```

## Map -- (Modify All)

```python
insurance_details["Criteria"] = list(map(lambda x: "Eligible" if x > 20 else "NA",insurance_details["age"]))
```

```python
insurance_details
```

## filter

```python
eligible_age=list(filter(lambda x:x>20,insurance_details["age"]))
eligible_age
```

## Reduce -- combine all into one

```python
from functools import reduce
total_charges=reduce(lambda x,y:x+y,insurance_details["charges"])
total_charges
```
